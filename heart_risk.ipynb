```python
x=df['health_risk']
y=df.drop("health_risk",axis=1)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(y,x,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.fit_transform(x_test)
```

```python
from tensorflow.keras.callbacks import EarlyStopping
import tensorflow
from tensorflow.keras.layers import Dropout
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import regularizers
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.layers import BatchNormalization
```

```python
model=Sequential([
    Dense(264,activation='relu',input_shape=(11,)),
    Dropout(0.3),
    Dense(128,activation='relu'),
    Dropout(0.3),
    Dense(1,activation='sigmoid')
])
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
his=model.fit(x_train,y_train,epochs=50,validation_data=(x_test,y_test),batch_size=32)
```

```python
# y_pred is probably output from model.predict()
y_pred_prob = model.predict(x_test)


y_pred = (y_pred_prob > 0.5).astype(int)

# Now import metrics and evaluate
from sklearn.metrics import classification_report
import warnings
warnings.filterwarnings('ignore')

print(classification_report(y_test, y_pred))
```
