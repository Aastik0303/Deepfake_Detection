## Load a pre-trained transformer model

**Reasoning**:
Load the tokenizer and model for a pre-trained transformer model suitable for summarization using the transformers library.



```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "t5-small"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("Tokenizer loaded:", tokenizer)
print("Model loaded:", model)
```

## Prepare the data

### Subtask:
Tokenize and encode the article and highlight texts using the tokenizer associated with the chosen pre-trained model. Ensure the data is in the correct format for the model, using PyTorch tensors.


**Reasoning**:
Define maximum lengths, tokenize and encode the article and highlight texts, convert them to PyTorch tensors, and create a PyTorch Dataset for training.



```python
import torch
from torch.utils.data import Dataset, DataLoader

max_article_length = 512
max_highlight_length = 128

class SummarizationDataset(Dataset):
    def __init__(self, articles, highlights, tokenizer, max_article_len, max_highlight_len):
        self.articles = articles
        self.highlights = highlights
        self.tokenizer = tokenizer
        self.max_article_len = max_article_len
        self.max_highlight_len = max_highlight_len

    def __len__(self):
        return len(self.articles)

    def __getitem__(self, idx):
        article = self.articles[idx]
        highlight = self.highlights[idx]

        article_encoding = self.tokenizer(
            article,
            max_length=self.max_article_len,
            padding='max_length',
            truncation=True,
            return_attention_mask=True,
            return_tensors='pt'
        )

        highlight_encoding = self.tokenizer(
            highlight,
            max_length=self.max_highlight_len,
            padding='max_length',
            truncation=True,
            return_attention_mask=True,
            return_tensors='pt'
        )

        return {
            'input_ids': article_encoding['input_ids'].squeeze(),
            'attention_mask': article_encoding['attention_mask'].squeeze(),
            'labels': highlight_encoding['input_ids'].squeeze()
        }

dataset = SummarizationDataset(df['article'].tolist(), df['highlights'].tolist(), tokenizer, max_article_length, max_highlight_length)

print(f"Number of samples in dataset: {len(dataset)}")
print(f"First sample input_ids shape: {dataset[0]['input_ids'].shape}")
print(f"First sample attention_mask shape: {dataset[0]['attention_mask'].shape}")
print(f"First sample labels shape: {dataset[0]['labels'].shape}")
```

## Fine-tune the model

### Subtask:
Fine-tune the pre-trained model on the prepared dataset for the summarization task using PyTorch. This involves training the model on the article-highlight pairs.


**Reasoning**:
Define the batch size, create a PyTorch DataLoader, set up the device, define the optimizer and number of epochs, and implement the training loop to fine-tune the model.



```python
batch_size = 8
dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

optimizer = torch.optim.AdamW(model.parameters(), lr=5e-5)

num_epochs = 3

print(f"Training on device: {device}")

for epoch in range(num_epochs):
    model.train()
    total_loss = 0
    for batch in dataloader:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['labels'].to(device)

        optimizer.zero_grad()

        outputs = model(input_ids=input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch+1}, Loss: {total_loss / len(dataloader)}")

print("Training finished.")
```

## Evaluate the model

### Subtask:
Evaluate the fine-tuned model using appropriate metrics for summarization, such as ROUGE scores.


**Reasoning**:
Import the necessary library for ROUGE scoring, create a test dataset and dataloader, and then iterate through the test data to generate summaries, calculate ROUGE scores, and accumulate them.



```python
!pip install rouge_score

from rouge_score import rouge_scorer
from torch.utils.data import DataLoader

scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)

test_dataset = SummarizationDataset(df['article'].tolist(), df['highlights'].tolist(), tokenizer, max_article_length, max_highlight_length)
test_dataloader = DataLoader(test_dataset, batch_size=batch_size)

model.eval()
total_rouge1 = 0
total_rouge2 = 0
total_rougel = 0
num_samples = 0

with torch.no_grad():
    for batch in test_dataloader:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['labels'].to(device)

        generated_ids = model.generate(
            input_ids=input_ids,
            attention_mask=attention_mask,
            max_length=max_highlight_length,
            num_beams=4,
            early_stopping=True
        )

        generated_summaries = tokenizer.batch_decode(generated_ids, skip_special_tokens=True)
        reference_highlights = tokenizer.batch_decode(labels, skip_special_tokens=True)

        for gen_summary, ref_highlight in zip(generated_summaries, reference_highlights):
            scores = scorer.score(ref_highlight, gen_summary)
            total_rouge1 += scores['rouge1'].fmeasure
            total_rouge2 += scores['rouge2'].fmeasure
            total_rougel += scores['rougeL'].fmeasure
            num_samples += 1

avg_rouge1 = total_rouge1 / num_samples
avg_rouge2 = total_rouge2 / num_samples
avg_rougel = total_rougel / num_samples

print(f"Average ROUGE-1: {avg_rouge1}")
print(f"Average ROUGE-2: {avg_rouge2}")
print(f"Average ROUGE-L: {avg_rougel}")
```

## Generate summaries

### Subtask:
Use the fine-tuned model to generate summaries for new articles.


**Reasoning**:
Select a few example articles, preprocess them, generate summaries using the fine-tuned model, decode the summaries, and print the original article and generated summary for each example.



```python
import torch

# Select a few example articles (using indices that were likely part of the dataset)
example_indices = [5, 10, 15]
example_articles = [df['article'][i] for i in example_indices]
example_highlights = [df['highlights'][i] for i in example_indices]

# Preprocess the example articles
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

preprocessed_articles = tokenizer(
    example_articles,
    max_length=max_article_length,
    padding='max_length',
    truncation=True,
    return_attention_mask=True,
    return_tensors='pt'
)

input_ids = preprocessed_articles['input_ids'].to(device)
attention_mask = preprocessed_articles['attention_mask'].to(device)

# Set the model to evaluation mode
model.eval()

# Generate summaries
with torch.no_grad():
    generated_ids = model.generate(
        input_ids=input_ids,
        attention_mask=attention_mask,
        max_length=max_highlight_length,
        num_beams=4,
        early_stopping=True
    )

# Decode the generated summaries
generated_summaries = tokenizer.batch_decode(generated_ids, skip_special_tokens=True)

# Print original articles and generated summaries
for i, (original_article, generated_summary, original_highlight) in enumerate(zip(example_articles, generated_summaries, example_highlights)):
    print(f"--- Example {i+1} ---")
    print("Original Article:")
    print(original_article)
    print("\nOriginal Highlight:")
    print(original_highlight)
    print("\nGenerated Summary:")
    print(generated_summary)
    print("-" * 20)
```
